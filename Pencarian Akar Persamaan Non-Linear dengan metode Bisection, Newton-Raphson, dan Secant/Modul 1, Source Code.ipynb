# BISECTION


```python
import math
def f(x):
  return 2 * math.sin(x) - math.exp(x/4) - 1

def bisection(f, a, b, tol=1e-06, max_iter=10):
  if f(a) * f(b) >= 0:
    print("Gagal")
    return None

  header_fmt = "{:<8}{:<12}{:<12}{:<12}{:<12}{}"
  row_fmt = "{:<8}{:<12.6f}{:<12.6f}{:<12.6f}{:<12.6f}{}"

  print(header_fmt.format("iterasi", "a", "b", "c", "f(c)", "interval baru"))
  print("-" * 70)

  for i in range(1, max_iter+1):
    c = (a + b)/2
    fc = f(c)
    a_lama, b_lama = a, b

    if  f(a)*f(c)<0:
      interval_baru = "[{:.6f};{:.6f}]". format(a_lama, b)
      b = c
    else:
      interval_baru = "[{:.6f};{:.6f}]". format(a, b_lama)
      a = c

    print(row_fmt.format(i, a_lama, b_lama, c, fc, interval_baru))

    if abs(fc) < tol or (b-a)/2 < tol:
      print("-" * 70)
      print("Konvergensi akar hampiran diperoleh pada iterasi ke-{}: x = {:.6f}". format(i, c))
      return c

  print("-" * 70)
  print("Mencapai iterasi maksimum {}. Akar hampiran terakhir: x = {:.6f}".format(max_iter, c))
  return c

a_awal = -5
b_awal = -3

bisection(f, a_awal, b_awal, tol=1e-06, max_iter=10)
```

# NEWTON RAPHSON

```python
import math
def f(x):
  return 2 * math.sin(x) - math.exp(x / 4) - 1
def df(x):
  return 2 * math.cos(x) -(1 / 4) * math.exp(x / 4)

def newton_rapshon(f, df, x0, tol=1e-06, max_iter=10):
  header_fmt = "{:<8}{:<12}{:<12}{:<12}{:<12}{}"
  row_fmt = "{:<8d}{:<12.6f}{:<12.6f}{:<12.6f}{:<12.6f}{:.6f}"
  print(header_fmt.format("iterasi", "x0", "f(x0)", "f'(x0)", "x1", "error"))
  print("-" * 70)

  x = x0

  for i in range(1, max_iter + 1):
     fx = f(x)
     dfx = df(x)

     if dfx == 0:
      print("-" * 70)
      print("Gagal: Turunan f'(x) = 0")
      return None

     x_baru = x - (fx / dfx)
     error = abs(x_baru - x)
     print(row_fmt.format(i, x, fx, dfx, x_baru, error))

     if error < tol:
       print("-" * 70)
       print("Konvergensi hampiran diperoleh pada iterasi ke-{}: x = {:.6f}".format(i, x_baru))
       return x_baru

     x = x_baru

  print("-" * 70)
  print("Mencapai iterasi maksimum {}. Akar hampiran terakhir: x = {:.6f}".format(max_iter, x))
  return x

x_awal = (-5 + (-3)) / 2
hasil = newton_rapshon(f, df, x_awal, tol=1e-06, max_iter=10)
print("-" * 70)
print("Hasil Akhir:", hasil)
```

# SECANT


```python
import math

def f(x):
    return 2 * math.sin(x) - math.exp(x / 4) - 1


def secant(f, x0, x1, tol=1e-6, max_iter=10):
    header_fmt = "{:<8}{:<12}{:<12}{:<12}{:<12}{:<12}"
    row_fmt = "{:<8}{:<12.6f}{:<12.6f}{:<12.6f}{:<12.6f}{:<12.6f}"

    print(header_fmt.format(
        "Iterasi", "x lama", "x baru", "f(x lama)", "f(x baru)", "Error"
    ))
    print("-" * 70)

    for i in range(1, max_iter + 1):
        fx0 = f(x0)
        fx1 = f(x1)

        if fx1 - fx0 == 0:
            print("Gagal: penyebut = 0")
            return None

        x2 = x1 - fx1 * (x1 - x0) / (fx1 - fx0)

        error = abs(x2 - x1)

        print(row_fmt.format(i, x0, x1, fx0, fx1, error))

        if abs(f(x2)) < tol or error < tol:
            print("-" * 70)
            print(
                "Konvergensi akar hampiran diperoleh "
                "pada iterasi ke-{}: x = {:.6f}".format(i, x2)
            )
            return x2

        x0 = x1
        x1 = x2

    print("-" * 70)
    print(
        "Mencapai iterasi maksimum {}."
        "Akar hampiran terakhir: x = {:.6f}".format(max_iter, x1)
    )

    return x1


x0 = -5
x1 = -3

hasil = secant(f, x0, x1, tol=1e-6, max_iter=10)

print("-" * 70)
print("Hasil akhir =", hasil)
```
